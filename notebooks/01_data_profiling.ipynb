```python
import os
import pandas as pd

pd.set_option("display.max_columns", None)
```

```python
data_path = "../data"

csv_files = sorted([f for f in os.listdir(data_path) if f.endswith(".csv")])

csv_files
```

```python
datasets = {}

for file in csv_files:
    datasets[file] = pd.read_csv(os.path.join(data_path, file))

print(f"{len(datasets)} tables loaded.")
```

```python
for name, df in datasets.items():
    print("="*60)
    print(name)
    print(df.shape)
```

```python
summary = []

for name, df in datasets.items():
    summary.append({
        "Table": name.replace("_dataset.csv", ""),
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

summary = pd.DataFrame(summary)

summary
```
